### Source and Format

**Dataset:** [Spotify Tracks Dataset](https://huggingface.co/datasets/maharshipandya/spotify-tracks-dataset)  
**File:** `spotify.csv`  
**Format:** CSV

Our data is stored in a file called spotify.csv, and it contains 114,000 songs with 114 different genres and 1,000 songs for each of the genres. Each row represents a song record, and the columns contain information about the song and its audio characteristics.
The dataset includes song information such as track ID, album name, track name, popularity, duration in milliseconds, and whether the lyrics are explicit. It also includes audio characteristics such as danceability, energy, key, loudness, mode, speechiness, acousticness, instrumentalness, liveness, valence, tempo, and time signature.

## Implementation/technical merit

1. Anticipated challenges:
   - Missing values: Some songs might be missing some of the audio characteristic attributes. To handle the missing values, we would decide to just remove those values.
   - Duplicate Songs: Songs might appear twice if they appear on multiple albums or under different artists. We would need to identify duplicates without accidentally removing remixes or other songs and deleting if they are duplicates.
   - Feature Scaling: The numerical features all have different ranges. Based on what classification algorithm we choose, these would need to be standardized.
   - Subjective classification: Genres are not always clearly defined. The dataset may assume the genre by the song's attributes.

2. Large Number of Attributes:
   - The dataset has a large number of attributes. In the process of building a model, we can determine which of those is helpful in predicting the genre and which is hurting to limit down the number to those that have an impact on classification.

## Potential impact of the results

1. These results are useful for understanding how to categorize songs into genres. By knowing the genre, it can help with sorting them into the correct playlist or the correct genre.

2. Some of the stakeholders could be Spotify users, music streaming platforms, and artists and producers. These groups could use the results to discover music, organize songs by genre, and understand how audio characteristics relate to different genres.
